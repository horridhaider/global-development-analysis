```python
import pandas as pd
import os

datasets = {
    "youth_mortality": "https://ourworldindata.org/grapher/youth-mortality-rate.csv?v=1&csvType=full&useColumnShortNames=false",
    
    "unemployment_rate": "https://ourworldindata.org/grapher/unemployment-rate.csv?v=1&csvType=full&useColumnShortNames=false",
    
    "share_in_poverty": "https://ourworldindata.org/explorers/poverty-explorer.csv?v=1&csvType=full&useColumnShortNames=false&Indicator=Share+in+poverty&Poverty+line=%243+per+day%3A+International+Poverty+Line&Household+survey+data+type=Show+data+from+both+income+and+consumption+surveys&Show+breaks+between+less+comparable+surveys=false",
    
    "number_in_poverty": "https://ourworldindata.org/explorers/poverty-explorer.csv?v=1&csvType=full&useColumnShortNames=false&Indicator=Number+in+poverty&Poverty+line=%243+per+day%3A+International+Poverty+Line&Household+survey+data+type=Show+data+from+both+income+and+consumption+surveys&Show+breaks+between+less+comparable+surveys=false",
    
    "population": "https://ourworldindata.org/explorers/population-and-demography.csv?v=1&csvType=full&useColumnShortNames=true&hideControls=false&indicator=Population&Sex=Both+sexes&Age=Total&Projection+scenario=None",
    
    "life_expectancy": "https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=false",
    
    "human_rights": "https://ourworldindata.org/grapher/human-rights-index-vdem.csv?v=1&csvType=full&useColumnShortNames=false",
    
    "homicide_rate": "https://ourworldindata.org/grapher/homicide-rate-unodc.csv?v=1&csvType=full&useColumnShortNames=false",
    
    "gdp": "https://ourworldindata.org/grapher/gdp-worldbank-constant-usd.csv?v=1&csvType=full&useColumnShortNames=false",
    
    "gdp_per_capita": "https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=false",
    
    "literacy_rate": "https://ourworldindata.org/grapher/cross-country-literacy-rates.csv?v=1&csvType=full&useColumnShortNames=false"
}

notebook_dir = os.path.abspath("")
parent_dir = os.path.abspath("..")

os.makedirs(os.path.join(parent_dir, "data"), exist_ok=True)
data_dir = os.path.join(parent_dir, "data")

dfs = {}

for name, url in datasets.items():
    df = pd.read_csv(
        url,
        storage_options={
            "User-Agent": "Our World In Data data fetch/1.0"
        }
    )
    
    dfs[name] = df
    df.to_csv(f"{data_dir}/{name}.csv", index=False)
    
    print(f"{name}: {df.shape}")
```
